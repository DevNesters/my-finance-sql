## Connect to Database

Establishes a connection to the PostgreSQL database using psycopg2 and the configuration from the previous cell. This connection will be used for all subsequent queries in the notebook.

## Database Configuration

Set your database connection parameters. These values should match your `config/database.yml` file. You can modify these values if your database credentials are different.

```python
DB_NAME = "her_finance_db" # @param {type:"string"}
DB_USER = "finance_user" # @param {type:"string"}
DB_PASSWORD = "her_finance_password" # @param {type:"string"}
DB_HOST = "localhost" # @param {type:"string"}
DB_PORT = 5432 # @param {type:"number"}
```

```python
# Import required libraries
import logging
from datetime import datetime

import pandas as pd
import plotly.express as px
import psycopg2

# Database connection
conn = psycopg2.connect(
    dbname=DB_NAME,
    user=DB_USER,
    password=DB_PASSWORD,
    host=DB_HOST,
    port=DB_PORT
)

print("Connected to database successfully!")
```

## Close Database Connection

Properly closes the database connection to free up resources. This is important to prevent connection leaks and ensures the cleanup script can drop the database if needed.

```python
conn.close()
print("Database connection closed.")
```
